```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))


inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

targets = [0, 1, 1, 0]

alpha = 0.8

x0 = 1

w13 = -0.2
w14 = 0.4
w23 = 0.2
w24 = -0.3
w35 = 0.2
w45 = -0.3

theta3 = 0.4
theta4 = 0.1
theta5 = -0.3


epochs = 10000


for epoch in range(epochs):

    total_error = 0

    for i in range(4):

        x1 = inputs[i][0]
        x2 = inputs[i][1]
        target = targets[i]

        I3 = x1 * w13 + x2 * w23 + x0 * theta3
        O3 = sigmoid(I3)

        I4 = x1 * w14 + x2 * w24 + x0 * theta4
        O4 = sigmoid(I4)

        I5 = O3 * w35 + O4 * w45 + x0 * theta5
        O5 = sigmoid(I5)

        Error5 = O5 * (1 - O5) * (target - O5)
        Error3 = O3 * (1 - O3) * Error5 * w35
        Error4 = O4 * (1 - O4) * Error5 * w45

        w13 = w13 + alpha * Error3 * x1
        w23 = w23 + alpha * Error3 * x2

        w14 = w14 + alpha * Error4 * x1
        w24 = w24 + alpha * Error4 * x2

        w35 = w35 + alpha * Error5 * O3
        w45 = w45 + alpha * Error5 * O4

        theta3 = theta3 + alpha * Error3
        theta4 = theta4 + alpha * Error4
        theta5 = theta5 + alpha * Error5

        total_error += abs(target - O5)

    if (epoch + 1) % 1000 == 0:
        print("Epoch:", epoch + 1,
              "Total Error:", round(total_error, 6))



print("\nFinal Weights and Biases")
print("--------------------------------")

print("w13 =", round(w13, 6))
print("w14 =", round(w14, 6))
print("w23 =", round(w23, 6))
print("w24 =", round(w24, 6))
print("w35 =", round(w35, 6))
print("w45 =", round(w45, 6))

print("theta3 =", round(theta3, 6))
print("theta4 =", round(theta4, 6))
print("theta5 =", round(theta5, 6))


print("\nXOR Results")
print("--------------------------------")

for i in range(4):

    x1 = inputs[i][0]
    x2 = inputs[i][1]


    I3 = x1 * w13 + x2 * w23 + theta3
    O3 = sigmoid(I3)

    I4 = x1 * w14 + x2 * w24 + theta4
    O4 = sigmoid(I4)

    I5 = O3 * w35 + O4 * w45 + theta5
    O5 = sigmoid(I5)

    print(
        "Input:", x1, x2,
        " Target:", targets[i],
        " Output:", round(O5, 4)
    )
```
