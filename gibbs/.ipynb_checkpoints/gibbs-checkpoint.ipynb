```python
import random


x = 0
y = 0

samples = []

for i in range(10):
    
    if y == 0:
        x = random.choice([0, 1])
    else:
        x = random.choice([1, 2])

    
    if x == 0:
        y = random.choice([0, 1])
    else:
        y = random.choice([1, 2])

    samples.append((x, y))

print("Gibbs Samples:")
for sample in samples:
    print(sample)
```
